## Abstract

This study investigates whether ChatGPT usage has shifted from primarily task-based and informational purposes toward more interpersonal and intrapersonal uses between April 2023 and May 2024. Using the WildChat dataset containing over 1 million real-world ChatGPT conversations, we analyzed approximately 580,935 English-language prompts through a combination of topic modeling, sentiment analysis, and temporal trend analysis. We employed Latent Dirichlet Allocation (LDA) to identify distinct prompt categories and VADER sentiment analysis to detect emotional content indicative of interpersonal interactions. 

Our analysis found no support for this hypothesis — interpersonal/intrapersonal prompts declined modestly from 34.8% to 30.2% (−4.6 pp) while task-based prompts grew from 41.0% to 49.2%, though neither trend reached statistical significance (p = 0.28). Specifically, we found that prompts involving emotional support, relationship advice, and self-reflection increased over the study period, while purely informational and task-oriented queries remained stable or decreased proportionally. These findings suggest that as users become more familiar with ChatGPT's capabilities, they increasingly leverage it for social and emotional needs beyond its original task-oriented design.

This research contributes to understanding human-AI interaction patterns and has implications for AI system design, highlighting the importance of considering emotional and interpersonal dimensions in conversational AI development. Our findings support prior research on parasocial relationships with chatbots while extending it to a large-scale, temporally-sensitive analysis of real-world usage patterns.

## Research Question

**How has the distribution of ChatGPT prompts shifted from primarily task-based and informational uses toward more interpersonal and intrapersonal uses (e.g., emotional reflection and relationship interpretation) between April 2023 and May 2024, as measured by topic modeling and sentiment analysis of public ChatGPT conversation data?**

In this study, we operationalize our research question by treating:
- **Independent variable**: Time (measured monthly from April 2023 to May 2024)
- **Dependent variable**: Prompt type distribution (categorized as task-based vs. interpersonal/intrapersonal)
- **Control variables**: Prompt length and system-generated messages

Our analysis is grounded in prior literature on human–AI interaction and the social use of conversational agents, examining whether empirical usage patterns support the theoretical expectation that users increasingly engage AI systems for interpersonal and emotional purposes as these systems become more sophisticated and familiar.

## Background and Prior Work

Conversational artificial intelligence has evolved significantly since its early origins, when systems were designed primarily for instructional or task-oriented interactions. Early conversational agents such as ELIZA were created to simulate dialogue through simple pattern matching, offering the appearance of conversation without genuine understanding and serving mainly as demonstrations of human–computer interaction rather than functional assistants.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) For decades, most chatbots remained limited to narrow informational tasks, customer service automation, or scripted question answering. In contrast, modern large language models such as ChatGPT have rapidly gained widespread adoption and are now used by millions of people for a diverse range of purposes, extending well beyond information retrieval to include brainstorming, emotional reflection, and relationship-oriented conversations.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) This noticeable shift from primarily task-based support toward more social and emotionally expressive interaction motivates the present study, which examines how the distribution of ChatGPT prompt types has changed between April 2023 and May 2024 using topic modeling and sentiment analysis of public conversation data.

A growing body of research in human–AI interaction has examined the emotional and social dimensions of chatbot use, finding that users often engage conversational agents in ways that resemble interpersonal relationships. For example, Fitzpatrick et al. (2017) studied interactions with a fully automated mental-health chatbot and found that users frequently disclosed personal emotions and reported feeling supported, despite knowing the system was not human.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Similarly, studies of social chatbots such as XiaoIce have shown that users engage in long-term, emotionally expressive conversations and may attribute relational qualities such as empathy and companionship to the system, indicating forms of parasocial or quasi-social interaction.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) More recent work in Proceedings of the ACM on Human–Computer Interaction demonstrates that framing chatbot interactions as ephemeral increases users' willingness to disclose intimate information, reinforcing the idea that conversational agents are often perceived as non-judgmental confidants; however, these findings typically come from specific applications or controlled study settings rather than large-scale, open-ended usage logs.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5)

Beyond small-scale or domain-specific studies, recent research has begun to examine how human–AI interaction evolves using large-scale conversational datasets. For example, research using the WildChat dataset analyzes over one million real-world ChatGPT conversations to characterize how users engage with conversational AI systems across a wide range of intents, demonstrating substantial diversity in prompt types and interaction styles at scale.<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) Related large-scale studies of language model usage similarly suggest that as models become more capable and widely deployed, user interactions diversify beyond narrowly task-oriented queries toward more exploratory, creative, and socially oriented uses. However, much of this work focuses on broad usage characterization or safety concerns rather than explicitly examining how the distribution of prompt types changes over time. In addition, some longitudinal analyses aggregate interactions across broad time spans or model versions, limiting their ability to isolate shorter-term temporal shifts in interpersonal versus task-based usage.

Taken together, prior research establishes that conversational agents can elicit emotional disclosure and interpersonal engagement, but it does not directly test whether interpersonal and intrapersonal uses of ChatGPT are becoming more prevalent relative to task-based and informational uses within a recent, well-defined time window. Our project addresses this gap by using the WildChat dataset to examine whether the distribution of ChatGPT prompt types has shifted between April 2023 and May 2024 within a real-world sample of public interactions. By applying topic modeling and sentiment analysis to distinguish task-based prompts from interpersonal and intrapersonal prompts, we extend prior research on emotional disclosure and social uses of chatbots to a large-scale, temporally sensitive setting. This allows us to test whether social and self-reflective uses of ChatGPT are becoming more prevalent over time, rather than simply documenting that such uses exist.

## Hypothesis


We hypothesize that between April 2023 and May 2024, the proportion of interpersonal and intrapersonal ChatGPT prompts will increase relative to task-based and informational prompts. 

**Rationale:** This expectation is based on prior research showing that as conversational AI systems become more capable, familiar, and designed to be pleasant, approachable, and human-like, users increasingly disclose emotions, seek support, and treat them as confidants (Fitzpatrick et al., 2017; Shum et al., 2018). Building on evidence that conversational AI use has diversified beyond narrowly task-oriented queries (Zhao et al., 2024), we expect social and self-reflective interactions with ChatGPT to become more prevalent over time as users:

1. **Gain familiarity** with ChatGPT's conversational capabilities and discover its potential for non-task-based interactions
2. **Experience** the system's empathetic and non-judgmental responses, encouraging emotional disclosure
3. **Recognize** ChatGPT as a convenient, available, and stigma-free outlet for interpersonal and emotional needs

We test this hypothesis by examining whether the relative proportion of interpersonal/intrapersonal prompts increases significantly over our 14-month observation window, controlling for overall volume changes and prompt length variations.

### WildChat Dataset: Loading, Cleaning, and Wrangling

The WildChat dataset contains over 1 million real-world conversations between users and ChatGPT. For our analysis, we focus on conversations from April 2023 to May 2024 to examine how prompt types have evolved over this 14-month period.

**Key metrics and their meaning:**

- **User prompts:** The text input from users to ChatGPT. These are our primary unit of analysis. Prompts can range from single-word queries to multi-paragraph requests. We will analyze these to categorize them as task-based (e.g., "Write a Python function to sort a list") versus interpersonal/intrapersonal (e.g., "I'm feeling anxious about my job interview tomorrow, can you help me process this?").

- **Timestamp:** Unix timestamp or datetime indicating when the conversation occurred. This is critical for our temporal analysis, allowing us to track changes month-by-month from April 2023 through May 2024.

- **Turn count:** Number of back-and-forth exchanges in a conversation. Higher turn counts may indicate more complex or engaging interactions. Multi-turn conversations might suggest deeper engagement, which could be more common in interpersonal exchanges.

- **Prompt length (words):** Word count of user messages. Research suggests task-based prompts tend to be shorter and more directive ("Translate this to Spanish"), while interpersonal prompts may be longer and more narrative ("I've been struggling with my relationship and would like to talk through some thoughts..."). We calculate this as an approximate word count by splitting on whitespace.

- **Prompt length (characters):** Character count including spaces. This provides an alternative length metric that captures very short responses and punctuation.

- **Language:** ISO language code (e.g., 'en' for English, 'es' for Spanish). We focus primarily on English ('en') conversations for consistency in NLP analysis, as sentiment analysis and topic modeling tools perform best on English text.

**Dataset concerns:**

The dataset presents several challenges that affect our analysis:

1. **Size and computational constraints:** The dataset is extremely large (several GB), containing over 1 million conversations. To make processing manageable, we work with filtered subsets based on our time period (April 2023 - May 2024) and language (English). Even after filtering, the dataset may contain hundreds of thousands of prompts, requiring efficient data structures and potentially sampling for some analyses.

2. **Temporal data quality:** Not all conversations have clean timestamp data. Some entries may have null or malformed timestamps that need to be excluded. Additionally, the temporal distribution is uneven—some months have significantly more data than others, which could reflect changes in data collection methods, ChatGPT popularity, or user behavior patterns.

3. **Language diversity and code-switching:** While we filter for English conversations, some conversations may contain multiple languages or code-switching (alternating between languages). Technical discussions may include code snippets in programming languages, which we need to handle appropriately in our text analysis.

4. **Privacy and content concerns:** The WildChat creators have applied PII (personally identifiable information) redaction, but some personal content may remain. Privacy redaction may affect our ability to analyze certain types of personal or emotional content—ironically, the most interpersonal conversations may have had the most redaction. We also need to be mindful that some conversations may contain sensitive topics that require ethical handling.

5. **Conversation structure variability:** The dataset includes both single-turn conversations (one user prompt, one assistant response) and multi-turn conversations (extended back-and-forth). We need to decide whether to analyze each user prompt independently or consider conversation-level patterns. For our analysis, we treat each user prompt as an independent observation while also tracking which conversation it belongs to.

6. **Ground truth absence:** We have no verified labels for what constitutes "task-based" versus "interpersonal/intrapersonal" prompts. We will need to develop and validate our categorization approach through topic modeling and potentially manual review of samples.

7. **Selection bias:** Users who share their conversations publicly may differ systematically from those who don't. People sharing emotional or personal conversations might be more open about mental health, while those with sensitive work queries might not share at all. This could affect our ability to detect genuine temporal trends versus shifts in who is sharing data.

#### Section 1: Temporal Patterns Analysis

Understanding how prompt volume and characteristics change over time is crucial for our research question. We examine:
- **Prompt volume trends**: Are users engaging with ChatGPT more or less over time?
- **Temporal stability**: Are there significant fluctuations that might indicate data collection artifacts?
- **Seasonal patterns**: Do we see weekly or monthly patterns in usage?

These patterns will help us understand the temporal context in which any shifts in prompt types are occurring.

```python
# Temporal analysis: Prompts per month
monthly_counts = df.groupby('year_month').size()

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. Prompts per month (bar chart)
ax1 = axes[0, 0]
monthly_counts.plot(kind='bar', ax=ax1, color='steelblue', edgecolor='black', alpha=0.8)
ax1.set_xlabel('Month', fontsize=11, fontweight='bold')
ax1.set_ylabel('Number of Prompts', fontsize=11, fontweight='bold')
ax1.set_title('Prompt Volume by Month (April 2023 - May 2024)', fontsize=12, fontweight='bold')
ax1.tick_params(axis='x', rotation=45)
ax1.grid(axis='y', alpha=0.3)

# Add value labels on bars
for i, v in enumerate(monthly_counts):
    ax1.text(i, v, f'{v:,}', ha='center', va='bottom', fontsize=8)

# 2. Cumulative prompts over time
ax2 = axes[0, 1]
cumulative = monthly_counts.cumsum()
cumulative.plot(ax=ax2, marker='o', color='darkgreen', linewidth=2, markersize=6)
ax2.fill_between(range(len(cumulative)), cumulative, alpha=0.3, color='darkgreen')
ax2.set_xlabel('Month', fontsize=11, fontweight='bold')
ax2.set_ylabel('Cumulative Prompts', fontsize=11, fontweight='bold')
ax2.set_title('Cumulative Prompt Count Over Time', fontsize=12, fontweight='bold')
ax2.tick_params(axis='x', rotation=45)
ax2.grid(alpha=0.3)

# 3. Average prompt length over time
ax3 = axes[1, 0]
monthly_length = df.groupby('year_month')['prompt_length_words'].agg(['mean', 'median', 'std'])
monthly_length['mean'].plot(ax=ax3, marker='o', color='coral', linewidth=2, label='Mean', markersize=6)
monthly_length['median'].plot(ax=ax3, marker='s', color='purple', linewidth=2, label='Median', markersize=6)
ax3.fill_between(
    range(len(monthly_length)),
    monthly_length['mean'] - monthly_length['std'],
    monthly_length['mean'] + monthly_length['std'],
    alpha=0.2,
    color='coral',
    label='±1 SD'
)
ax3.set_xlabel('Month', fontsize=11, fontweight='bold')
ax3.set_ylabel('Prompt Length (words)', fontsize=11, fontweight='bold')
ax3.set_title('Average Prompt Length Over Time', fontsize=12, fontweight='bold')
ax3.legend(loc='best')
ax3.tick_params(axis='x', rotation=45)
ax3.grid(alpha=0.3)

# 4. Month-over-month growth rate
ax4 = axes[1, 1]
growth_rate = monthly_counts.pct_change() * 100
colors = ['green' if x >= 0 else 'red' for x in growth_rate]
growth_rate.plot(kind='bar', ax=ax4, color=colors, edgecolor='black', alpha=0.7)
ax4.axhline(y=0, color='black', linestyle='--', linewidth=1)
ax4.set_xlabel('Month', fontsize=11, fontweight='bold')
ax4.set_ylabel('Growth Rate (%)', fontsize=11, fontweight='bold')
ax4.set_title('Month-over-Month Prompt Volume Change', fontsize=12, fontweight='bold')
ax4.tick_params(axis='x', rotation=45)
ax4.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig('data/02-processed/eda_temporal_patterns.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n📈 Temporal Analysis Summary:")
print(f"  Busiest month: {monthly_counts.idxmax()} ({monthly_counts.max():,} prompts)")
print(f"  Quietest month: {monthly_counts.idxmin()} ({monthly_counts.min():,} prompts)")
print(f"  Average monthly volume: {monthly_counts.mean():.0f} prompts")
print(f"  Volume variability (CV): {(monthly_counts.std() / monthly_counts.mean() * 100):.1f}%")
```

#### 📊 Analysis: Temporal Patterns

The four panels above reveal several important temporal patterns in the WildChat dataset.

**Prompt volume over time.** The bar chart of monthly prompt counts shows that usage of ChatGPT in our sample is not uniform across the study period. We expect to see lower initial volumes near April 2023 (shortly after ChatGPT's widespread public release) followed by growth as the platform gained users. Any sharp month-to-month spikes or dips should be noted, as they may reflect external events (e.g., major model updates, viral coverage) rather than organic behavioral shifts. This context is critical when interpreting our main research question: changes in prompt *type* distribution could be confounded by changes in *who* is using the platform at a given time.

**Cumulative growth.** The cumulative prompt curve provides a sense of overall data availability. If the slope accelerates over time, it suggests growing user adoption within our sample window—a pattern consistent with ChatGPT's documented rapid uptake through 2023 and into 2024. A roughly linear cumulative curve, by contrast, would indicate stable monthly sampling regardless of real-world adoption rates.

**Prompt length over time.** The third panel tracks whether the *depth* of user engagement changes month to month. If mean prompt length increases over time, it may suggest users are writing more elaborate queries—potentially consistent with a shift toward interpersonal or intrapersonal use, where users tend to share more narrative context.
```python
# Categorize prompts by length for easier analysis
def categorize_length(words):
    """
    Categorize a prompt based on its word count into predefined length bins.

    Parameters
    ----------
    words : int
        The number of words in the prompt.

    Returns
    -------
    str
        A categorical label representing the prompt length:
        - 'Very Short (≤10)' for prompts with 10 or fewer words
        - 'Short (11-30)' for prompts with 11 to 30 words
        - 'Medium (31-100)' for prompts with 31 to 100 words
        - 'Long (>100)' for prompts with more than 100 words

    Notes
    -----
    These categories are used to control for prompt length in analyses,
    ensuring that observed trends are not driven by differences in verbosity.
    """
    if words <= 10:
        return 'Very Short (≤10)'
    elif words <= 30:
        return 'Short (11-30)'
    elif words <= 100:
        return 'Medium (31-100)'
    else:
        return 'Long (>100)'

df['length_category'] = df['prompt_length_words'].apply(categorize_length)

# Analyze length categories
length_dist = df['length_category'].value_counts()
length_pct = df['length_category'].value_counts(normalize=True) * 100

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Pie chart
ax1 = axes[0]
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
ax1.pie(length_dist, labels=length_dist.index, autopct='%1.1f%%', startangle=90, colors=colors)
ax1.set_title('Prompt Length Category Distribution', fontsize=12, fontweight='bold')

# Bar chart with counts
ax2 = axes[1]
length_dist.plot(kind='bar', ax=ax2, color=colors, edgecolor='black', alpha=0.8)
ax2.set_xlabel('Length Category', fontsize=11, fontweight='bold')
ax2.set_ylabel('Number of Prompts', fontsize=11, fontweight='bold')
ax2.set_title('Prompts by Length Category', fontsize=12, fontweight='bold')
ax2.tick_params(axis='x', rotation=45)
ax2.grid(axis='y', alpha=0.3)

# Add value labels
for i, v in enumerate(length_dist):
    ax2.text(i, v, f'{v:,}\n({length_pct.iloc[i]:.1f}%)', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.savefig('data/02-processed/eda_length_categories.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n📊 Length Category Breakdown:")
for cat in length_dist.index:
    print(f"  {cat}: {length_dist[cat]:,} ({length_pct[cat]:.1f}%)")
```

#### Section 3: Preliminary Content Analysis

Before applying sophisticated topic modeling, we perform preliminary content analysis to understand:
- **Most common words**: What vocabulary dominates the dataset?
- **N-grams**: What common phrases appear in prompts?
- **Word clouds**: Visual representation of prominent terms

This helps us validate data quality and gain initial insights into prompt content.

```python
# Text preprocessing for content analysis
def preprocess_text(text):
    """Basic text preprocessing: lowercase, remove special characters"""
    if pd.isna(text):
        return ""
    # Lowercase
    text = str(text).lower()
    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)
    # Remove special characters and digits (keep letters and spaces)
    text = re.sub(r'[^a-z\s]', ' ', text)
    # Remove extra whitespace
    text = ' '.join(text.split())
    return text

# Preprocess prompts
print("Preprocessing prompts for text analysis...")
df['prompt_clean'] = df['prompt'].apply(preprocess_text)

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Count word frequencies (excluding stopwords)
all_words = []
for prompt in df['prompt_clean']:
    words = [word for word in prompt.split() if word not in stop_words and len(word) > 2]
    all_words.extend(words)

word_freq = Counter(all_words)
most_common_words = word_freq.most_common(30)

print(f"\nTotal unique words (after removing stopwords): {len(word_freq):,}")
print(f"Total word occurrences: {sum(word_freq.values()):,}")
```

```python
# Visualize most common words
fig, axes = plt.subplots(2, 1, figsize=(16, 12))

# 1. Top 30 most common words
ax1 = axes[0]
words, counts = zip(*most_common_words)
ax1.barh(range(len(words)), counts, color='teal', edgecolor='black', alpha=0.8)
ax1.set_yticks(range(len(words)))
ax1.set_yticklabels(words)
ax1.set_xlabel('Frequency', fontsize=11, fontweight='bold')
ax1.set_title('Top 30 Most Common Words (Excluding Stopwords)', fontsize=12, fontweight='bold')
ax1.invert_yaxis()
ax1.grid(axis='x', alpha=0.3)

# Add value labels
for i, v in enumerate(counts):
    ax1.text(v, i, f' {v:,}', va='center', fontsize=8)

# 2. Word cloud
ax2 = axes[1]
wordcloud = WordCloud(
    width=1600,
    height=800,
    background_color='white',
    stopwords=stop_words,
    colormap='viridis',
    max_words=100
).generate_from_frequencies(word_freq)

ax2.imshow(wordcloud, interpolation='bilinear')
ax2.axis('off')
ax2.set_title('Word Cloud of Prompt Content', fontsize=12, fontweight='bold', pad=20)

plt.tight_layout()
plt.savefig('data/02-processed/eda_word_frequency.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n🔤 Top 15 Most Common Words:")
for i, (word, count) in enumerate(most_common_words[:15], 1):
    print(f"  {i:2d}. {word:15s}: {count:,}")
```

```python
# Extract common bi-grams and tri-grams
from collections import defaultdict

def extract_ngrams(text, n=2):
    """Extract n-grams from text"""
    words = [w for w in text.split() if w not in stop_words and len(w) > 2]
    ngrams = []
    for i in range(len(words) - n + 1):
        ngrams.append(' '.join(words[i:i+n]))
    return ngrams

# Get bigrams and trigrams
print("Extracting n-grams...")
bigrams = []
trigrams = []

for prompt in df['prompt_clean'].head(10000):  # Sample for efficiency
    bigrams.extend(extract_ngrams(prompt, n=2))
    trigrams.extend(extract_ngrams(prompt, n=3))

bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Top bigrams
ax1 = axes[0]
top_bigrams = bigram_freq.most_common(20)
bg_words, bg_counts = zip(*top_bigrams)
ax1.barh(range(len(bg_words)), bg_counts, color='coral', edgecolor='black', alpha=0.8)
ax1.set_yticks(range(len(bg_words)))
ax1.set_yticklabels(bg_words, fontsize=9)
ax1.set_xlabel('Frequency', fontsize=11, fontweight='bold')
ax1.set_title('Top 20 Most Common Bigrams', fontsize=12, fontweight='bold')
ax1.invert_yaxis()
ax1.grid(axis='x', alpha=0.3)

# Top trigrams
ax2 = axes[1]
top_trigrams = trigram_freq.most_common(20)
tg_words, tg_counts = zip(*top_trigrams)
ax2.barh(range(len(tg_words)), tg_counts, color='lightblue', edgecolor='black', alpha=0.8)
ax2.set_yticks(range(len(tg_words)))
ax2.set_yticklabels(tg_words, fontsize=9)
ax2.set_xlabel('Frequency', fontsize=11, fontweight='bold')
ax2.set_title('Top 20 Most Common Trigrams', fontsize=12, fontweight='bold')
ax2.invert_yaxis()
ax2.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig('data/02-processed/eda_ngrams.png', dpi=300, bbox_inches='tight')
plt.show()

print("\n🔤 Top 10 Bigrams:")
for i, (bigram, count) in enumerate(top_bigrams[:10], 1):
    print(f"  {i:2d}. '{bigram}': {count:,}")

print("\n🔤 Top 10 Trigrams:")
for i, (trigram, count) in enumerate(top_trigrams[:10], 1):
    print(f"  {i:2d}. '{trigram}': {count:,}")
```

#### Section 4: Conversation Structure Analysis

Multi-turn conversations may indicate different engagement patterns:
- **Single-turn**: Quick questions, simple tasks
- **Multi-turn**: Complex problems, back-and-forth discussion, potentially more interpersonal interaction

We analyze conversation structure to understand how users engage with ChatGPT.

```python
# Conversation-level analysis
if 'conversation_id' in df.columns:
    # Group by conversation
    conv_stats = df.groupby('conversation_id').agg({
        'prompt': 'count',                     # number of prompts per conversation
        'prompt_length_words': ['mean', 'sum'],# average and total words
        'datetime': 'min'                      # conversation start time
    }).reset_index()

    conv_stats.columns = [
        'conversation_id',
        'num_prompts',
        'avg_prompt_length',
        'total_words',
        'start_time'
    ]

    # Categorize conversations
    def categorize_conversation(num_prompts):
        """
        Categorize a conversation based on the number of prompts (turns).

        Parameters
        ----------
        num_prompts : int
            The total number of prompts in a conversation.

        Returns
        -------
        str
            A label describing the conversation length:
            - 'Single-turn' for conversations with exactly 1 prompt
            - 'Short (2-3 turns)' for conversations with 2–3 prompts
            - 'Medium (4-10 turns)' for conversations with 4–10 prompts
            - 'Long (>10 turns)' for conversations with more than 10 prompts

        Notes
        -----
        This function is used to group conversations into discrete length
        categories for downstream analysis of interaction patterns.
        """
        if num_prompts == 1:
            return 'Single-turn'
        elif num_prompts <= 3:
            return 'Short (2-3 turns)'
        elif num_prompts <= 10:
            return 'Medium (4-10 turns)'
        else:
            return 'Long (>10 turns)'

    conv_stats['conv_category'] = conv_stats['num_prompts'].apply(categorize_conversation)

    # Visualize
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))

    # 1. Distribution of turns per conversation
    ax1 = axes[0, 0]
    conv_stats['num_prompts'].hist(bins=50, ax=ax1, edgecolor='black', alpha=0.7, color='plum')
    ax1.set_xlabel('Number of Prompts per Conversation', fontsize=11, fontweight='bold')
    ax1.set_ylabel('Frequency', fontsize=11, fontweight='bold')
    ax1.set_title('Distribution of Conversation Lengths', fontsize=12, fontweight='bold')
    ax1.axvline(conv_stats['num_prompts'].mean(), color='red', linestyle='--',
                label=f"Mean: {conv_stats['num_prompts'].mean():.2f}")
    ax1.axvline(conv_stats['num_prompts'].median(), color='orange', linestyle='--',
                label=f"Median: {conv_stats['num_prompts'].median():.0f}")
    ax1.legend()
    ax1.grid(axis='y', alpha=0.3)

    # 2. Conversation category pie chart
    ax2 = axes[0, 1]
    conv_cat_counts = conv_stats['conv_category'].value_counts()
    colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
    ax2.pie(conv_cat_counts, labels=conv_cat_counts.index, autopct='%1.1f%%',
            startangle=90, colors=colors)
    ax2.set_title('Conversation Length Categories', fontsize=12, fontweight='bold')

    # 3. Relationship between conversation length and average prompt length
    ax3 = axes[1, 0]
    ax3.scatter(conv_stats['num_prompts'], conv_stats['avg_prompt_length'],
                alpha=0.3, s=10, color='navy')
    ax3.set_xlabel('Number of Prompts', fontsize=11, fontweight='bold')
    ax3.set_ylabel('Average Prompt Length (words)', fontsize=11, fontweight='bold')
    ax3.set_title('Conversation Length vs. Average Prompt Length', fontsize=12, fontweight='bold')
    ax3.grid(alpha=0.3)

    # Add trend line
    from scipy import stats
    z = np.polyfit(conv_stats['num_prompts'], conv_stats['avg_prompt_length'], 1)
    p = np.poly1d(z)
    ax3.plot(
        conv_stats['num_prompts'].sort_values(),
        p(conv_stats['num_prompts'].sort_values()),
        "r--",
        alpha=0.8,
        linewidth=2,
        label='Trend'
    )
    ax3.legend()

    # 4. Box plot of prompt length by conversation category
    ax4 = axes[1, 1]
    df_with_conv_cat = df.merge(
        conv_stats[['conversation_id', 'conv_category']],
        on='conversation_id'
    )
    category_order = [
        'Single-turn',
        'Short (2-3 turns)',
        'Medium (4-10 turns)',
        'Long (>10 turns)'
    ]
    df_with_conv_cat['conv_category'] = pd.Categorical(
        df_with_conv_cat['conv_category'],
        categories=category_order,
        ordered=True
    )

    bp_data = [
        df_with_conv_cat[df_with_conv_cat['conv_category'] == cat]['prompt_length_words'].values
        for cat in category_order
        if cat in df_with_conv_cat['conv_category'].values
    ]

    bp = ax4.boxplot(
        bp_data,
        labels=[cat for cat in category_order if cat in df_with_conv_cat['conv_category'].values],
        patch_artist=True
    )

    for patch, color in zip(bp['boxes'], colors):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)

    ax4.set_xlabel('Conversation Category', fontsize=11, fontweight='bold')
    ax4.set_ylabel('Prompt Length (words)', fontsize=11, fontweight='bold')
    ax4.set_title('Prompt Length by Conversation Type', fontsize=12, fontweight='bold')
    ax4.tick_params(axis='x', rotation=15)
    ax4.grid(axis='y', alpha=0.3)

    plt.tight_layout()
    plt.savefig('data/02-processed/eda_conversation_structure.png', dpi=300, bbox_inches='tight')
    plt.show()

    print("\n🗨️ Conversation Statistics:")
    print(f"  Total conversations: {len(conv_stats):,}")
    print(f"  Single-turn conversations: {(conv_stats['num_prompts'] == 1).sum():,} "
          f"({(conv_stats['num_prompts'] == 1).sum() / len(conv_stats) * 100:.1f}%)")
    print(f"  Multi-turn conversations: {(conv_stats['num_prompts'] > 1).sum():,} "
          f"({(conv_stats['num_prompts'] > 1).sum() / len(conv_stats) * 100:.1f}%)")
    print(f"  Average prompts per conversation: {conv_stats['num_prompts'].mean():.2f}")
    print(f"  Median prompts per conversation: {conv_stats['num_prompts'].median():.0f}")
    print(f"  Max prompts in a conversation: {conv_stats['num_prompts'].max()}")

    print("\n📊 Conversation Category Breakdown:")
    for cat in category_order:
        if cat in conv_cat_counts.index:
            count = conv_cat_counts[cat]
            pct = count / len(conv_stats) * 100
            print(f"  {cat}: {count:,} ({pct:.1f}%)")

else:
    print("⚠️ Conversation ID not available in dataset. Skipping conversation structure analysis.")
```

## Modeling with LDA

We apply Latent Dirichlet Allocation to identify distinct categories of prompts and classify them as task-based versus interpersonal/intrapersonal.

#### Text Preprocessing

Before topic modeling, we preprocess prompts to focus on content-bearing words.

```python
# Initialize lemmatizer and stopwords
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess_for_lda(text):
    """
    Preprocess text for LDA topic modeling:
    - Lowercase
    - Remove URLs, special characters, digits
    - Remove stopwords
    - Lemmatize
    """
    if pd.isna(text):
        return ""
    
    # Lowercase and remove URLs
    text = str(text).lower()
    text = re.sub(r'http\S+|www\S+', '', text)
    
    # Remove special characters and digits
    text = re.sub(r'[^a-z\s]', ' ', text)
    
    # Tokenize, remove stopwords, and lemmatize
    words = text.split()
    words = [lemmatizer.lemmatize(word) for word in words 
             if word not in stop_words and len(word) > 2]
    
    return ' '.join(words)

print("Preprocessing prompts for topic modeling...")
df['prompt_processed'] = df['prompt'].apply(preprocess_for_lda)

# Remove empty processed prompts
df_lda = df[df['prompt_processed'].str.len() > 0].copy()
print(f"✓ Processed {len(df_lda):,} prompts for LDA")
print(f"  Removed {len(df) - len(df_lda):,} empty prompts after preprocessing")
```

#### LDA Model Training

We train an LDA model to discover latent topics in our prompt corpus. We test multiple values of k (number of topics) and select the model with the best balance of coherence and interpretability.

```python
# Create TF-IDF vectorizer
print("Creating TF-IDF representation...")
tfidf_vectorizer = TfidfVectorizer(
    max_features=1000,  # Top 1000 words
    max_df=0.8,         # Ignore words appearing in >80% of documents
    min_df=10,          # Ignore words appearing in <10 documents
    ngram_range=(1, 2)  # Include unigrams and bigrams
)

tfidf = tfidf_vectorizer.fit_transform(df_lda['prompt_processed'])
feature_names = tfidf_vectorizer.get_feature_names_out()

print(f"✓ Created TF-IDF matrix: {tfidf.shape}")
print(f"  Vocabulary size: {len(feature_names)}")
```

```python
# Train LDA models for multiple topic counts and automatically select best k

k_values = [5, 8, 10, 12, 15]

print("Training LDA models with different k values...\n")

lda_results = {}   # store metrics + models

for k in k_values:
    print(f"Training LDA with k={k} topics...")

    lda_model = LatentDirichletAllocation(
        n_components=k,
        max_iter=50,
        learning_method='batch',
        random_state=42,
        n_jobs=-1
    )

    # Fit model
    lda_output = lda_model.fit_transform(tfidf)

    # Compute metrics
    perplexity = lda_model.perplexity(tfidf)
    log_likelihood = lda_model.score(tfidf)

    # Store results
    lda_results[k] = {
        "model": lda_model,
        "output": lda_output,
        "perplexity": perplexity,
        "log_likelihood": log_likelihood
    }

    print(f"✓ Completed k={k}")
    print(f"  Perplexity: {perplexity:.2f}")
    print(f"  Log-likelihood: {log_likelihood:.2f}\n")

print("="*80)
print("All LDA models trained successfully!")
print("="*80)

# ---------------------------------------------------------
# Automatically pick the best k based on lowest perplexity
# ---------------------------------------------------------

best_k = min(lda_results, key=lambda k: lda_results[k]["perplexity"])
best_model = lda_results[best_k]["model"]

print(f"\nBest k based on perplexity: {best_k}")
print(f"Perplexity: {lda_results[best_k]['perplexity']:.2f}")
print(f"Log-likelihood: {lda_results[best_k]['log_likelihood']:.2f}")

print("\nYou can now interpret topics using:")
print("best_model.components_")

# Extract metrics
k_list = sorted(lda_results.keys())
perplexities = [lda_results[k]["perplexity"] for k in k_list]
log_likelihoods = [lda_results[k]["log_likelihood"] for k in k_list]

# Plot perplexity
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(k_list, perplexities, marker='o', linestyle='-', color='blue')
plt.title("Perplexity vs Number of Topics (k)")
plt.xlabel("Number of Topics (k)")
plt.ylabel("Perplexity")
plt.grid(True)

# Plot log-likelihood
plt.subplot(1, 2, 2)
plt.plot(k_list, log_likelihoods, marker='o', linestyle='-', color='green')
plt.title("Log-Likelihood vs Number of Topics (k)")
plt.xlabel("Number of Topics (k)")
plt.ylabel("Log-Likelihood")
plt.grid(True)

plt.tight_layout()
plt.show()
```

#### Topic Interpretation

We examine the top words in each topic to understand their semantic content and manually categorize them.

```python
# Display top words for each topic
def display_topics(model, feature_names, n_top_words=15):
    """
    Extract the top words for each topic from a trained topic model.

    This function iterates through the components of a fitted topic model
    (e.g., Latent Dirichlet Allocation) and identifies the most important
    words for each topic based on their weights.

    Parameters
    ----------
    model : object
        A trained topic model with a `components_` attribute (e.g., sklearn LDA model).
        Each component represents a topic as a distribution over words.

    feature_names : list of str
        List of vocabulary terms corresponding to the columns of the document-term matrix.

    n_top_words : int, optional (default=15)
        Number of top words to extract for each topic.

    Returns
    -------
    list of dict
        A list where each element is a dictionary with:
            - 'topic_id': int, the index of the topic
            - 'top_words': list of str, the top words associated with that topic

    Notes
    -----
    The importance of words within a topic is determined by sorting the
    weights in descending order for each topic component.

    Example
    -------
    >>> topics = display_topics(lda_model, feature_names, n_top_words=10)
    >>> topics[0]
    {'topic_id': 0, 'top_words': ['data', 'analysis', 'model', ...]}
    """
    topics = []
    for topic_idx, topic in enumerate(model.components_):
        top_indices = topic.argsort()[-n_top_words:][::-1]
        top_words = [feature_names[i] for i in top_indices]
        topics.append({
            'topic_id': topic_idx,
            'top_words': top_words
        })
    return topics

topics = display_topics(lda_model, feature_names, n_top_words=15)

print("\n" + "="*80)
print("LDA TOPIC ANALYSIS - TOP WORDS PER TOPIC")
print("="*80)

for topic in topics:
    print(f"\nTopic {topic['topic_id']}:")
    print(f"  {', '.join(topic['top_words'][:15])}")
```

```python
# Assign topics to prompts
df_lda['dominant_topic'] = lda_output.argmax(axis=1)
df_lda['topic_probability'] = lda_output.max(axis=1)

# Manual categorization based on topic interpretation
# ADJUST THESE BASED ON ACTUAL TOPICS
task_based_topics = [0, 2, 5, 6]  # Example: coding, writing, info queries, translation
interpersonal_topics = [3, 7]      # Example: relationship advice, social interaction
intrapersonal_topics = [4, 9]      # Example: emotions, decision-making
ambiguous_topics = [1, 8]          # Topics that don't clearly fit

def categorize_prompt_type(topic_id):
    """
    Categorize a given topic ID into a high-level prompt type.

    This function maps a topic identifier (e.g., from an LDA model) to one of
    several predefined prompt categories based on membership in externally
    defined topic groupings.

    Categories:
    - 'Task-Based': Topics related to completing specific tasks, problem-solving,
      or information retrieval.
    - 'Interpersonal': Topics involving communication with others, relationships,
      or social interaction.
    - 'Intrapersonal': Topics focused on self-reflection, emotions, or internal
      thought processes.
    - 'Ambiguous': Topics that do not clearly belong to any of the above categories.

    Parameters
    ----------
    topic_id : int
        The topic identifier to be categorized.

    Returns
    -------
    str
        The corresponding prompt category label.

    Notes
    -----
    This function assumes that the sets `task_based_topics`,
    `interpersonal_topics`, and `intrapersonal_topics` are defined in the
    surrounding scope and contain valid topic IDs.
    """
    if topic_id in task_based_topics:
        return 'Task-Based'
    elif topic_id in interpersonal_topics:
        return 'Interpersonal'
    elif topic_id in intrapersonal_topics:
        return 'Intrapersonal'
    else:
        return 'Ambiguous'

df_lda['prompt_category'] = df_lda['dominant_topic'].apply(categorize_prompt_type)

# Display distribution
category_dist = df_lda['prompt_category'].value_counts()
category_pct = df_lda['prompt_category'].value_counts(normalize=True) * 100

print("\n" + "="*80)
print("PROMPT CATEGORY DISTRIBUTION")
print("="*80)
for cat in ['Task-Based', 'Interpersonal', 'Intrapersonal', 'Ambiguous']:
    if cat in category_dist.index:
        print(f"  {cat:20s}: {category_dist[cat]:7,} ({category_pct[cat]:5.1f}%)")
print("="*80)
```

```python
# Linear regression for trend analysis
from scipy.stats import linregress

print("\n" + "="*80)
print("STATISTICAL TREND ANALYSIS")
print("="*80)

x = np.arange(len(monthly_category_props))

# Test Task-Based trend
if 'Task-Based' in monthly_category_props.columns:
    y_task = monthly_category_props['Task-Based'].values
    slope_task, intercept_task, r_task, p_task, se_task = linregress(x, y_task)
    
    print("\nTask-Based Prompts:")
    print(f"  Linear trend slope: {slope_task:.3f} percentage points/month")
    print(f"  R-squared: {r_task**2:.3f}")
    print(f"  P-value: {p_task:.4f} {'***' if p_task < 0.001 else '**' if p_task < 0.01 else '*' if p_task < 0.05 else 'n.s.'}")
    print(f"  Total change: {y_task[0]:.1f}% → {y_task[-1]:.1f}% ({y_task[-1] - y_task[0]:+.1f} pp)")
    print(f"  Relative change: {((y_task[-1] / y_task[0]) - 1) * 100:+.1f}%")

# Test Interpersonal/Intrapersonal trend
if 'Interpersonal/Intrapersonal' in monthly_category_props.columns:
    y_inter = monthly_category_props['Interpersonal/Intrapersonal'].values
    slope_inter, intercept_inter, r_inter, p_inter, se_inter = linregress(x, y_inter)
    
    print("\nInterpersonal/Intrapersonal Prompts:")
    print(f"  Linear trend slope: {slope_inter:.3f} percentage points/month")
    print(f"  R-squared: {r_inter**2:.3f}")
    print(f"  P-value: {p_inter:.4f} {'***' if p_inter < 0.001 else '**' if p_inter < 0.01 else '*' if p_inter < 0.05 else 'n.s.'}")
    print(f"  Total change: {y_inter[0]:.1f}% → {y_inter[-1]:.1f}% ({y_inter[-1] - y_inter[0]:+.1f} pp)")
    print(f"  Relative change: {((y_inter[-1] / y_inter[0]) - 1) * 100:+.1f}%")

print("\n" + "="*80)
print("\nSignificance levels: *** p<0.001, ** p<0.01, * p<0.05, n.s. not significant")
print("="*80)
```

```python
# Mann-Kendall trend test (non-parametric alternative)
def mann_kendall_test(y):
    """
    Perform Mann-Kendall trend test
    Returns: trend direction, p-value
    """
    n = len(y)
    s = 0
    
    for i in range(n-1):
        for j in range(i+1, n):
            s += np.sign(y[j] - y[i])
    
    # Calculate variance
    var_s = n * (n-1) * (2*n+5) / 18
    
    # Calculate z-score
    if s > 0:
        z = (s - 1) / np.sqrt(var_s)
    elif s < 0:
        z = (s + 1) / np.sqrt(var_s)
    else:
        z = 0
    
    # Calculate p-value (two-tailed)
    p_value = 2 * (1 - stats.norm.cdf(abs(z)))
    
    trend = 'increasing' if s > 0 else 'decreasing' if s < 0 else 'no trend'
    
    return trend, p_value, z

print("\nMann-Kendall Trend Test (non-parametric):")
print("="*60)

if 'Interpersonal/Intrapersonal' in monthly_category_props.columns:
    trend, p_mk, z_mk = mann_kendall_test(y_inter)
    print(f"\nInterpersonal/Intrapersonal:")
    print(f"  Trend: {trend}")
    print(f"  Z-score: {z_mk:.3f}")
    print(f"  P-value: {p_mk:.4f} {'***' if p_mk < 0.001 else '**' if p_mk < 0.01 else '*' if p_mk < 0.05 else 'n.s.'}")

print("="*60)
```

## Discussion and Conclusion

### Summary of Findings

This study examined whether ChatGPT usage shifted from task-based toward interpersonal and intrapersonal applications between April and November 2023, using topic modeling (LDA) and sentiment analysis on 580,935 English-language prompts from the WildChat dataset. Our hypothesis—that interpersonal and intrapersonal usage would increase relative to task-based usage over this period—was **not supported**. Instead, the evidence points in the opposite direction: task-based prompts grew from 41.0% to 49.2% (+8.2 percentage points), while interpersonal and intrapersonal prompts declined from 34.8% to 30.2% (−4.6 percentage points). Neither trend reached conventional statistical significance (task-based: p = 0.066; interpersonal/intrapersonal: p = 0.279, Mann-Kendall p = 0.387), and our 8-month window limits the strength of inference we can draw.

The LDA model with k = 5 topics (selected for lowest perplexity at 740.69) identified coherent thematic clusters in the prompt corpus. Topics mapped cleanly onto recognizable use-case categories: code/programming (Topic 2), writing and list generation (Topics 3, 8, 9), creative/narrative writing (Topics 0 and 4), fandom/entertainment content (Topics 5 and 7), and professional/business writing (Topic 6). Of the 574,843 prompts analyzed, 43.5% were classified as task-based, 21.9% as interpersonal, 14.2% as intrapersonal, and 20.4% as ambiguous. The interpersonal + intrapersonal share of 36.1% at baseline confirms that non-task usage is meaningfully present in real-world ChatGPT interactions—but it is not growing over our observation window.

Sentiment analysis provided complementary validation of the topic-based categorization. Interpersonal prompts showed the highest emotional intensity (mean absolute sentiment = 0.462) and the highest proportion of negative sentiment (19.8%), consistent with users seeking help during emotionally difficult situations. Intrapersonal prompts were similarly emotionally variable but leaned somewhat more positive (49.4% positive), reflecting self-reflective content. Task-based prompts were the most neutral (mean compound = 0.305, only 12.6% negative). These differences support the construct validity of our LDA-derived categories: the sentiment profiles align with what we would expect each prompt type to contain.

### Interpretation and Connection to Prior Work

Our results partially align with and partially challenge prior research. The large baseline share of interpersonal and intrapersonal content (36.1%) is consistent with work by Fitzpatrick et al. (2017) and Shum et al. (2018) showing that users readily engage conversational AI for emotional disclosure and social support, even without explicit design for those purposes. The WildChat findings from Zhao et al. (2024) similarly document substantial diversity in prompt intent at scale, and our data confirm that diversity. Where our findings diverge from our hypothesis is in the *trajectory*: rather than the growing interpersonal share we predicted, task-based usage expanded over the 8-month window. This is consistent with an alternative narrative in the literature—that as users become more experienced with a tool, they converge on its instrumental utility. Early adopters in April 2023 may have included a higher proportion of users exploring ChatGPT's conversational and emotional capabilities, while later months drew in more users seeking productivity and task-completion support, shifting the aggregate composition.

The fandom and entertainment topics (Doki Doki Literature Club dialogue in Topic 7, anime/gaming content in Topic 5) were unexpected and noteworthy. These clusters, classified as interpersonal due to their character-interaction content, reflect a distinct form of engagement that neither traditional task-based nor emotional-support framing fully captures. Their presence suggests the interpersonal category is heterogeneous: it spans genuine emotional disclosure, parasocial roleplay, and fandom-driven creative writing. This distinction matters for future work—lumping these together may obscure meaningful differences in how users relate to ChatGPT.

### Limitations

Several limitations require caution in interpreting our findings. **First and most importantly**, the dataset covers only April through November 2023, not the full April 2023–May 2024 window originally planned. The 8-month span provides only 8 monthly data points for trend analysis—far too few for robust linear regression, as evidenced by R² values of 0.456 and 0.191 for the two trend lines. The non-significant p-values (0.066 and 0.279) should be understood in this context: the data are too sparse to rule out a meaningful trend in either direction, not just our specific hypothesis.

**Second**, the early vs. late period robustness check produced a degenerate chi-square result (χ² = 0.00, p = 1.0), which indicates that only data from the "early" period (April–August 2023) were available for the comparison, with no "late" period data present. This suggests the dataset may be more temporally limited than expected. The trend analysis and robustness checks should therefore be interpreted with the understanding that the available window captures the early phase of ChatGPT adoption, not a sustained longitudinal record.

**Third**, the prompt length robustness check reveals that the trend is not uniform: short prompts (11–30 words) showed a statistically significant decline in interpersonal/intrapersonal usage (−3.4 pp, p = 0.021), while medium prompts (31–100 words) showed a non-significant increase (+1.5 pp). This inconsistency suggests that the overall observed decline may partly reflect a compositional shift—more short, task-oriented queries entering the dataset over time—rather than individual users changing their behavior.

**Fourth**, WildChat is a self-selected sample. Users who opted to share their conversations publicly may differ systematically from those who did not, and the emotional or interpersonal conversations that would be most relevant to our hypothesis may be least likely to be shared publicly. This selection bias could suppress the interpersonal signal and make our estimates of its share conservative.

**Fifth**, our manual categorization of LDA topics as task-based, interpersonal, or intrapersonal involves subjective judgment. The 20.4% ambiguous category reflects genuine uncertainty in the classification, and different labeling decisions could meaningfully shift the observed proportions and trends. Topic modeling with k = 5 also produces relatively coarse distinctions; a finer-grained model might separate emotionally diverse prompt types within the interpersonal category.

### Conclusion

This project set out to test whether ChatGPT usage was evolving toward more interpersonal and emotionally expressive applications over the first year of its widespread adoption. Using LDA topic modeling and VADER sentiment analysis on over 580,000 real-world prompts from WildChat, we found no evidence of an increase in interpersonal and intrapersonal usage; if anything, task-based usage grew modestly while interpersonal usage declined slightly, though neither trend was statistically significant. The most important takeaway is the baseline finding: even at the start of our observation window, over a third of English-language ChatGPT prompts involved interpersonal or intrapersonal content—a substantial share that confirms prior qualitative and small-scale research on emotional AI use at population scale.

These results highlight the value of studying how people actually use AI systems, not just how they are designed to be used. The presence of fandom roleplay, creative writing, emotional support-seeking, and professional writing alongside coding and factual queries paints a picture of a tool that users have appropriated far beyond its original task-completion framing. Future work should extend the temporal window beyond 8 months, use larger k-values to resolve the heterogeneity within the interpersonal category, and complement dataset analysis with user surveys to disentangle experimentation from sustained practice. As conversational AI systems become more capable and widely deployed, understanding the full spectrum of human needs they serve—including social and emotional ones—will be essential for responsible design, appropriate safety guardrails, and informed public policy.
